```
import numpy as np
import pandas as pd
import librosa
import librosa.display
import matplotlib.pyplot as plt
import IPython.display as ipd
```

Загрузка и анализ аудио сигналов

```
audio_file = '/1.opus'
signal, sr = librosa.load(audio_file)
```

```
# посмотрим тип нашей переменный signal

print(type(signal), type(sr))
print(signal.shape, sr)
```

```
# плеер

ipd.Audio(audio_file)
```

```
# визулизация аудиосигнала

plt.figure(figsize=(12,4))
librosa.display.waveshow(signal, sr=sr)
```

```
# частота - цветность

plt.figure(figsize=(12,4))
x = librosa.stft(signal) # разбивка аудиосигнала
xdb= librosa.amplitude_to_db(abs(x)) # переводим в ДЦб
librosa.display.specshow(xdb, sr=sr, x_axis='time', y_axis='hz')
plt.colorbar()
```
